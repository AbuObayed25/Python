***Breadth First Search***


*   Uninformed search
*   priority ,Semantic analysis
*   path searching algorithm
*   root- left - right





```python
graph = {
    '5':['3','7'],
    '3':['2','4'],
    '7':['8'],
    '2':[],
    '4':['8'],
    '8':[]

}
visited = []#List for visited nodes
queue = []#Initialize a queue
```

```python
def bfs(visited,graph,node):
  visited.append(node)
  queue.append(node)
  while queue:
    m = queue.pop(0)
    print(m,end=" ")
    for neighbour in graph[m]:
      if neighbour not in visited:
        visited.append(neighbour)
        queue.append(neighbour)
```

```python
print("Following is the Breadth-First Search")
bfs(visited,graph,'5')
```

Problem:
Suppose Alice have a country land Car parking Company where people can hire car for hiking .Now cars are occupied by the different state depending on the weather .Now Apply BFS to find out the priority of available cars with connected roots on the given dataset.

Dataset:
1. Cars:Car1 ....Car6
2. Weather:Car4 is ocoupied for 4 hour for NY & CA city and Car rest of them is for 1 hour.
3. city: Boston ,CA,Mexico,NY,Kansas,AB.

```python
from collections import deque
available_car = {
    "Car1": 1,
    "Car2": 1,
    "Car3": 1,
    "Car4": 4,
    "Car5": 1,
    "Car6": 1
}

graph = {
    "Car1": ["Boston"],
    "Car2": ["CA"],
    "Car3": ["Mexico"],
    "Car4": ["NY", "CA"],
    "Car5": ["Kansas"],
    "Car6": ["AB"]
}

def bfs_order(start):
    visited = set()
    queue = deque([start])
    order = []
    while queue:
        car = queue.popleft()
        if car not in visited:
            visited.add(car)
            order.append(car)
            for city in graph[car]:
                if city not in visited:
                    queue.append(city)
    return order

start_car = "Car4"
order = bfs_order(start_car)
print(order)
```
